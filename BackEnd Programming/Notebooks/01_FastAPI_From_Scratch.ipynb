# FastAPI From Scratch
Build a small typed HTTP API, validate requests, and test the contract.


```python
# Install in a notebook environment if needed
# %pip install fastapi uvicorn httpx

from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI(title='Learning API')

class Item(BaseModel):
    name: str
    quantity: int

@app.post('/items')
async def create_item(item: Item):
    return {'name': item.name, 'quantity': item.quantity}
```

```python
from fastapi.testclient import TestClient

client = TestClient(app)
response = client.post('/items', json={'name':'book','quantity':2})
print(response.status_code, response.json())
```
