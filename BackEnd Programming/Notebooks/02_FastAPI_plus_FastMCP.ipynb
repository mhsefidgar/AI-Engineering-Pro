# FastAPI + FastMCP
Connect a normal application API with an MCP server. The two interfaces serve different purposes: HTTP APIs expose application contracts, while MCP exposes tools/resources/prompts to MCP-aware clients.


```python
# %pip install fastapi fastmcp
from fastapi import FastAPI

api = FastAPI(title='Application API')

@api.get('/health')
def health():
    return {'status': 'ok'}
```

```python
from fastmcp import FastMCP

mcp = FastMCP('learning-tools')

@mcp.tool
def lookup_order(order_id: str) -> dict:
    # Replace this with a repository/service call in a real application.
    return {'order_id': order_id, 'status': 'processing'}
```
