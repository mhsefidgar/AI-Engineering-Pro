# AI API Backend
Build the backend boundary around an LLM provider without putting provider calls directly in route handlers.


```python
import os
from dataclasses import dataclass

@dataclass
class AIResult:
    output: str
    input_tokens: int = 0
    output_tokens: int = 0

class AIProvider:
    async def generate(self, prompt: str) -> AIResult:
        raise NotImplementedError
```

## Backend responsibilities
Validate input before the provider call. Apply authentication, rate limits, timeouts, retries, and cost controls. Keep the provider adapter behind a service boundary. Return a stable application schema rather than leaking provider-specific response objects.


```python
def estimate_cost(input_tokens: int, output_tokens: int, input_price: float, output_price: float) -> float:
    return (input_tokens / 1_000_000) * input_price + (output_tokens / 1_000_000) * output_price

print(estimate_cost(1000, 500, 1.0, 2.0))
```
