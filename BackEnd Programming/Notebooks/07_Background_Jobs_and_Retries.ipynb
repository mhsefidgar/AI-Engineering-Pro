# Background jobs, retries, and idempotency

Model slow or unreliable work as durable jobs rather than blocking an HTTP request.

```python
from dataclasses import dataclass
from datetime import datetime, timezone

@dataclass
class Job:
    id: str
    status: str = 'queued'
    attempts: int = 0
    available_at: datetime = datetime.now(timezone.utc)
```

```python
def retry_delay(attempt: int, base: float = 1.0, cap: float = 60.0) -> float:
    # Add random jitter in a production implementation.
    return min(cap, base * (2 ** max(0, attempt - 1)))
```
