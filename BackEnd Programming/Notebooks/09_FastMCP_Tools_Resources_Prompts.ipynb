# FastMCP: tools, resources, prompts, and safe execution

Build an MCP server around an existing service boundary. MCP should expose capabilities; domain logic remains reusable by REST and other callers.

```python
# %pip install fastmcp
from fastmcp import FastMCP

mcp = FastMCP('backend-lab')

@mcp.tool
async def search_projects(query: str) -> list[dict]:
    if not query.strip():
        raise ValueError('query is required')
    return [{'id': 1, 'name': 'demo-project'}]
```

```python
@mcp.resource('project://{project_id}')
async def project(project_id: int) -> str:
    return f'Project {project_id}'

@mcp.prompt
async def summarize_project(project_id: int) -> str:
    return f'Summarize authorized project {project_id} data and identify missing evidence.'
```
