# AI API streaming, budgets, and observability

Treat an LLM call as an unreliable, metered dependency rather than a single `generate()` function.

```python
from dataclasses import dataclass

@dataclass(frozen=True)
class Usage:
    input_tokens: int
    output_tokens: int

def estimate_cost(usage: Usage, input_price: float, output_price: float) -> float:
    return usage.input_tokens / 1_000_000 * input_price + usage.output_tokens / 1_000_000 * output_price
```

```python
async def stream_chunks(provider):
    async for chunk in provider:
        # A production API should emit framed events and a final usage event.
        yield chunk
```
