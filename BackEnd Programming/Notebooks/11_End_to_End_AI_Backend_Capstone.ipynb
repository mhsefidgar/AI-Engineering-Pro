# End-to-end AI backend capstone

Build the full path: authenticated FastAPI -> service layer -> PostgreSQL -> Redis -> queue -> RAG -> LLM -> optional FastMCP tools -> observability.

## Required API

`POST /v1/documents` starts ingestion.

`GET /v1/jobs/{id}` reports ingestion status.

`POST /v1/answers` performs retrieval + generation and returns citations, usage, request ID, and model metadata.

`GET /health` and `GET /ready` support operations.

## Implementation milestones

1. Define OpenAPI and data models.
2. Add JWT/OIDC authentication and tenant authorization.
3. Add PostgreSQL migrations and repository layer.
4. Add Redis caching/rate limits.
5. Add durable ingestion jobs.
6. Add chunking/embedding/vector retrieval.
7. Add provider-neutral LLM adapter and streaming.
8. Add optional FastMCP tools using the same service layer.
9. Add structured logs, metrics, tracing, and request IDs.
10. Add Docker and CI.

## Required tests

Unit: domain/service logic.

Integration: database, Redis, queue, and provider adapters.

API: authentication, authorization, validation, pagination, idempotency, errors.

AI: retrieval quality, citation correctness, token/cost limits, provider failures.

Security: object-level authorization, prompt injection, SSRF where applicable, secret leakage.

Reliability: dependency timeout, retries, duplicate jobs, worker crash, graceful shutdown.

## Production definition of done

A fresh developer can run the stack from documented commands; secrets are externalized; migrations are reproducible; tests pass in CI; health/readiness checks work; failures are observable; expensive AI operations have budgets/rate limits; data access is tenant-safe; and the architecture diagram matches the implementation.
